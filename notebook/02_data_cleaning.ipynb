# Data Cleaning

```python
import pandas as pd
```

## 1. Leyendo el dataframe

```python
df = pd.read_csv('../data/raw/social_media_vs_productivity.csv')
df
```

## 2. Rellenando datos nulos

```python
# Vemos el porcentaje de datos nulos que hay en cada columna
df_nulls = df.isnull().sum().reset_index()
df_nulls.rename(columns={0 : 'count'}, inplace = True)
df_nulls['count'] = (df_nulls['count'] / df_nulls['count'].sum() * 100).round(2)
df_nulls
```

```python
# Hacemos una función para rellenar los valores nulos a partir de combinaciones de columnas importantes
def fillnancombinations(column : str):

    """
    Fill NaN values in the specified column using the mean of groups defined by categorical columns.
    """

    return df[column].fillna(
        df.groupby(['gender', 'job_type', 'social_platform_preference'])[column].transform('mean')
    )
```

```python
## Rellenando las columnas con datos nulos
df['daily_social_media_time'] = fillnancombinations('daily_social_media_time')

df['perceived_productivity_score'] = fillnancombinations('perceived_productivity_score')

df['actual_productivity_score'] = fillnancombinations('actual_productivity_score')

df['stress_level'] = fillnancombinations('stress_level')

df['sleep_hours'] = fillnancombinations('sleep_hours')

df['screen_time_before_sleep'] = fillnancombinations('screen_time_before_sleep')

df['job_satisfaction_score'] = fillnancombinations('job_satisfaction_score')
```
